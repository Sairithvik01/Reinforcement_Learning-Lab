```python
import gymnasium as gym

def run_random_agent(env_name, num_episodes):
    # 1. Create the selected environment
    env = gym.make(env_name)

    all_episode_rewards = []

    print(f"\n{'='*40}")
    print(f"--- Running Random Agent in {env_name} ---")
    print(f"{'='*40}")

    for episode in range(1, num_episodes + 1):
        # 2. Reset the environment
        # 3. Observe the initial state
        state, info = env.reset()

        episode_reward = 0
        done = False

        # 8. Continue until the episode terminates
        while not done:
            # 4. Identify the available actions using the environment's action space
            # 5. Select an action randomly
            action = env.action_space.sample()

            # 6. Execute the action in the environment
            # 7. Observe the next state and reward
            next_state, reward, terminated, truncated, info = env.step(action)

            episode_reward += reward

            # The episode ends if it is terminated (e.g., pole falls) or truncated (time limit)
            done = terminated or truncated

        all_episode_rewards.append(episode_reward)
        print(f"Episode {episode}: Total Reward = {episode_reward}")

    # 9. Calculate and display the total reward and average reward obtained
    total_reward_across_episodes = sum(all_episode_rewards)
    average_reward = total_reward_across_episodes / num_episodes

    print(f"\n--- Final Results for {env_name} ---")
    print(f"Overall Total Reward: {total_reward_across_episodes}")
    print(f"Average Reward per Episode: {average_reward:.2f}\n")

    env.close()

if __name__ == "__main__":
    # Input section
    try:
        episodes_input = input("Enter the number of episodes to run for each environment: ")
        num_episodes = int(episodes_input)
    except ValueError:
        print("Invalid input. Defaulting to 3 episodes.")
        num_episodes = 3

    # Environments required by the assignment
    environments = ["CartPole-v1", "MountainCar-v0", "FrozenLake-v1"]

    # Run the random agent on all required environments
    for env in environments:
        run_random_agent(env, num_episodes)
```
