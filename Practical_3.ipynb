```python
import random

def main():
    # Part A - Machine Initialization
    machines = ['A', 'B', 'C']

    # Part B - Create knowledge (Initialize N, R, and Q tables)
    N = {m: 0 for m in machines}
    Q = {m: 0.0 for m in machines}
    R_total = {m: 0.0 for m in machines} # To track sum of rewards per machine

    # Given parameters
    n_play = 10
    epsilon = 0.2
    total_player_reward = 0.0

    print("Starting Modified Epsilon-Greedy Multi-Armed Bandit")
    print(f"Machines: {machines}, Total Plays: {n_play}, Epsilon: {epsilon}\n")

    for play in range(1, n_play + 1):
        print(f"--- Play {play} ---")

        # 1. Generate a random number r between 0 and 1
        r = random.random()

        # Identify the machine with the highest current Q-value
        max_q = max(Q.values())
        best_machines = [m for m, q in Q.items() if q == max_q]
        best_machine = random.choice(best_machines) # Break ties randomly

        # Part C - Modified epsilon-Greedy Decision
        if r < epsilon:
            # 2. Exploration: Randomly select one except the highest estimated reward machine
            available_for_exploration = [m for m in machines if m != best_machine]
            # Fallback if all machines are somehow excluded (e.g., all tied and we removed all)
            if not available_for_exploration:
                selected_machine = random.choice(machines)
            else:
                selected_machine = random.choice(available_for_exploration)
            print(f"r = {r:.4f} < {epsilon} -> Exploration (Excluded Machine {best_machine})")
        else:
            # 3. Exploitation: Select the machine having the highest current Q-value
            selected_machine = best_machine
            print(f"r = {r:.4f} >= {epsilon} -> Exploitation (Selected Best Machine)")

        # Part D - Reward Collection
        print(f"Play with machine {selected_machine}")
        reward_input = float(input("Enter reward: "))

        # Part E - Q-Value Update
        # 5. Update the number of times the selected machine has been played
        N[selected_machine] += 1
        R_total[selected_machine] += reward_input
        total_player_reward += reward_input

        # 6. Update its Q-value using the incremental sample-average formula
        # Q_new = Q_old + (1/N) * (Reward - Q_old)
        Q[selected_machine] = Q[selected_machine] + (1.0 / N[selected_machine]) * (reward_input - Q[selected_machine])

        print(f"Updated Q-value for {selected_machine}: {Q[selected_machine]:.4f}\n")

    # Part F - Display Results
    print("=== Final Results (N, R, Q Tables) ===")
    print(f"{'Machine':<10} | {'Plays (N)':<10} | {'Total R':<10} | {'Q-Value':<10}")
    print("-" * 47)
    for m in machines:
        print(f"{m:<10} | {N[m]:<10} | {R_total[m]:<10} | {Q[m]:.4f}")

    print(f"\nTotal reward the player received: {total_player_reward}")

if __name__ == "__main__":
    main()
```
