```python
import math

def ucb_multi_armed_bandit():
    # Part A: Machine Initialization
    machines = ['A', 'B', 'C']
    N_plays = 10

    # Initialize Tables
    N = {m: 0 for m in machines}
    Q = {m: 0.0 for m in machines}
    UCB = {m: 0.0 for m in machines}

    total_reward = 0
    current_play = 1

    print("--- Part B: Initial Test (Knowledge Creation) ---")
    # Play each machine once to avoid division by zero in UCB calculation
    for m in machines:
        print(f"\nPlay {current_play}: Selected machine {m}")
        reward = float(input(f"Enter reward for machine {m}: "))

        # Update N, R(Total), and Q
        N[m] += 1
        total_reward += reward
        # Incremental sample-average formula
        Q[m] = Q[m] + (1 / N[m]) * (reward - Q[m])
        current_play += 1

    print("\n--- Part C, D, E: UCB Strategy ---")
    # Proceed with the remaining plays up to N_plays
    while current_play <= N_plays:
        # 1. Calculate UCB for each machine
        # Formula: UCB(a) = Q(a) + sqrt((2 * ln(t)) / N(a))
        t = current_play - 1 # Total plays completed so far

        for m in machines:
            exploration_term = math.sqrt((2 * math.log(t)) / N[m])
            UCB[m] = Q[m] + exploration_term

        # 2. Choose machine with highest UCB value
        selected_machine = max(UCB, key=UCB.get)

        print(f"\nPlay {current_play}:")
        print(f"UCB Values -> A: {UCB['A']:.3f}, B: {UCB['B']:.3f}, C: {UCB['C']:.3f}")
        print(f"Selected machine: {selected_machine} (Highest UCB)")

        # 3. Obtain reward from user
        reward = float(input(f"Enter reward for machine {selected_machine}: "))
        total_reward += reward

        # 4 & 5. Update N and Q-value
        N[selected_machine] += 1
        Q[selected_machine] = Q[selected_machine] + (1 / N[selected_machine]) * (reward - Q[selected_machine])

        current_play += 1

    # Part F: Display Results
    print("\n--- Part F: Final Results ---")
    print(f"{'Machine':<10} | {'N (Plays)':<10} | {'Q (Estimated Value)':<20} | {'Final UCB':<10}")
    print("-" * 60)
    for m in machines:
        print(f"{m:<10} | {N[m]:<10} | {Q[m]:<20.3f} | {UCB[m]:.3f}")

    print(f"\nTotal Reward Accumulated: {total_reward}")

if __name__ == "__main__":
    ucb_multi_armed_bandit()
```
