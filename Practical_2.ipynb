```python
import random
machines = ['A', 'B', 'C']
N_play = 10
epsilon = 0.2
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]
total_reward = 0
for play in range(N_play):
    r = random.random()
    if r < epsilon:
        index = random.randint(0, 2)
        print("Exploration")
    else:
        index = Q.index(max(Q))
        print("Exploitation")
    machine = machines[index]
    print("\nPlay", play + 1)
    print("Machine selected:", machine)
    reward = float(input("Enter reward: "))
    N[index] = N[index] + 1
    R[index] = R[index] + reward
    Q[index] = Q[index] + (reward - Q[index]) / N[index]
    total_reward = total_reward + reward
    print("N =", N)
    print("R =", R)
    print("Q =", Q)
print("\n========== FINAL RESULT ==========")
for i in range(3):
    print(
        "Machine", machines[i],
        ": N =", N[i],
        ", R =", R[i],
        ", Q =", round(Q[i], 2)
    )
print("\nTotal Reward after 10 plays =", total_reward)
```
